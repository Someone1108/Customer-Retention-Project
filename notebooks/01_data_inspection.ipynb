# 01 - Data Inspection

In this notebook, we will inspect the raw Telco Customer Churn dataset before cleaning or modelling it.

The goal is not to transform everything immediately. The goal is to understand what we have:

- How many rows and columns are in the dataset?
- What does one customer record look like?
- Which columns are numeric, categorical, or identifiers?
- Are there missing values or suspicious fields?
- Which fields might become dimensions or facts later?


## 1. Import Libraries

`pandas` helps us load and inspect tabular data.

`Path` helps us build file paths in a cleaner way than typing long strings everywhere.

```python
from pathlib import Path

import pandas as pd
```

## 2. Load the Raw Dataset

We keep the original file in `data/raw` so we always have an untouched source file to return to.

```python
raw_file = Path("../data/raw/Telco_customer_churn.xlsx")

df = pd.read_excel(raw_file)

df.head()
```

## 3. Check Dataset Size

`shape` returns `(rows, columns)`.

For this project, each row should represent one customer.

```python
df.shape
```

## 4. Review Column Names

Column names tell us what business concepts are available in the dataset.

Later, these columns will help us decide which fields belong in dimension tables and which belong in fact tables.

```python
df.columns.tolist()
```

## 5. Inspect Data Types

Data types matter because Power BI, SQL, and Python treat numbers, text, and dates differently.

For example, if a revenue column is stored as text, calculations may break or behave strangely.

```python
df.info()
```

**Your notes:**

- Which columns look like customer identifiers?
- Which columns look like revenue or numeric measure fields?
- Are any numeric-looking fields stored as text?

## 6. Check Missing Values

Missing values are one of the first data quality checks.

This gives us a column-by-column count of blanks or nulls.

```python
missing_values = df.isna().sum().sort_values(ascending=False)

missing_values[missing_values > 0]
```

## 7. Check Duplicate Customers

A customer-level dataset should usually have one row per `CustomerID`.

If duplicate customer IDs exist, we need to investigate before building metrics.

```python
df["CustomerID"].duplicated().sum()
```

## 8. Explore Churn Distribution

Before dashboarding, we should understand the basic churn split.

```python
df["Churn Label"].value_counts(dropna=False)
```

```python
df["Churn Label"].value_counts(normalize=True, dropna=False).mul(100).round(2)
```

- What is the approximate churn rate?
- Is churn a minority or majority outcome?
- Why does this matter for retention analysis?
